### Core Analysis

#### Total runs per match

```python
# runs_per_match = final_df.groupby('match_id')['total_runs'].sum()


runs_per_match = (
    final_df
    .groupby(['match_id', 'team1', 'team2'])['total_runs']
    .sum()
    .reset_index(name='total_runs_per_match')
)
runs_per_match = runs_per_match.sort_values('total_runs_per_match', ascending=False)
```

```python
runs_per_match.head(10)
```

#### Runs per team per match

```python
runs_per_team = (final_df.groupby(['match_id', 'batting_team'])['total_runs'].sum().reset_index(name='runs'))
runs_per_team.head(5)
```

#### Top 10 batters

```python
runs_per_batsman = (final_df.groupby('batter')['batsman_runs'].sum().reset_index(name = 'total_runs'))
runs_per_batsman
```

```python
top_batsman = (runs_per_batsman.sort_values('total_runs', ascending=False).reset_index(drop = True))

top_batsman.insert(0, 'rank', top_batsman.index+1)
top_batsman.head(10)
```

#### Strike Rate of Batters

```python
runs_per_batsman = (final_df.groupby('batter')['batsman_runs'].sum())
runs_per_batsman
```

```python
balls_faced_per_batsman = (final_df[final_df['extras_type'] != 'wides'].groupby('batter').size())
balls_faced_per_batsman
```

```python
batsman_stats = pd.concat(
    [runs_per_batsman, balls_faced_per_batsman],
    axis=1
)

batsman_stats.columns = ['total_runs', 'balls_faced']
```

```python
batsman_stats
```

```python
batsman_stats['strike_rate'] = (batsman_stats['total_runs']/batsman_stats['balls_faced'])*100
```

```python
batsman_stats.head()
```

```python
batsman_stats = batsman_stats.sort_values('strike_rate', ascending=False)
```

```python
batsman_stats.head(10)
```

#### Top 10 bowlers by Economy

```python
runs_conceded = (final_df.groupby('bowler')['total_runs'].sum())
runs_conceded
```

```python
balls_bowled = (final_df[final_df['extra_runs'] == 0].groupby('bowler').size())
balls_bowled
```

```python
overs_bowled = balls_bowled / 6
```

```python
bowler_stats = pd.concat([runs_conceded, overs_bowled], axis = 1)

bowler_stats.columns = ['runs_conceded', 'overs_bowled']

bowler_stats['economy'] = bowler_stats['runs_conceded']/bowler_stats['overs_bowled']
bowler_stats = bowler_stats.sort_values('economy')
```

```python
bowler_stats.head(10)
```

#### Boundary Percentage

```python
total_runs = (
    final_df
    .groupby('batter')['batsman_runs']
    .sum()
    .reset_index(name='total_runs')
)
```

```python
total_runs
```

```python
boundary_runs = (
    final_df[final_df['batsman_runs'].isin([4, 6])]
    .groupby('batter')['batsman_runs']
    .sum()
    .reset_index(name='boundary_runs')
)
```

```python
boundary_runs
```

```python
runs_summary = (
    total_runs
    .merge(boundary_runs, on='batter', how='left')
    .fillna(0)
)

runs_summary['boundary_percentage'] = (
    runs_summary['boundary_runs'] / runs_summary['total_runs'] * 100
).round(2)

runs_summary.head(20)
```

#### Toss Impact Analysis

```python
innings_runs = (
    final_df
    .groupby(['match_id', 'inning', 'batting_team'])['total_runs']
    .sum()
    .reset_index(name='innings_runs')
)
```

```python
innings_runs
```

```python
innings_runs = innings_runs.merge(
    matches_df[['id', 'toss_winner']],
    left_on='match_id',
    right_on='id',
    how='left'
)
```

```python
innings_runs.head()
```

```python
innings_runs['team_type'] = innings_runs.apply(
    lambda x: 'toss_winner'
    if x['batting_team'] == x['toss_winner']
    else 'opponent',
    axis=1
)

innings_runs.head()
```

```python
toss_comparison = (
    innings_runs
    .groupby('team_type')['innings_runs']
    .mean()
    .reset_index(name='avg_runs')
)

toss_comparison
```

Yes winning toss has an advntage, as teams end up scoring more and win the match by winning the toss
